```python
# ==============================================================
# Notebook 1: Load and Prepare the Greece Earthquake Dataset
#
# Dataset:
# earthquakes_greece_2016_2026.csv
#
# Goal:
# Load the data, inspect it, clean it, and prepare it for
# exploratory analysis.
# ==============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Make plots look nicer
plt.style.use("ggplot")

# --------------------------------------------------------------
# 1. Load the dataset
# --------------------------------------------------------------

df = pd.read_csv("earthquakes_greece_2016_2026.csv")

# --------------------------------------------------------------
# 2. Inspect the data
# --------------------------------------------------------------

print("First five rows:")
print(df.head())

print("\nDataset information:")
print(df.info())

print("\nColumns:")
print(df.columns.tolist())

# --------------------------------------------------------------
# 3. Convert timestamps
# --------------------------------------------------------------

# Convert the USGS time column to datetime
df["time"] = pd.to_datetime(df["time"])

# Sort chronologically
df = df.sort_values("time").reset_index(drop=True)

# --------------------------------------------------------------
# 4. Keep only useful columns
# --------------------------------------------------------------

df = df[[
    "time",
    "latitude",
    "longitude",
    "depth",
    "mag",
    "place"
]]

# --------------------------------------------------------------
# 5. Remove missing values
# --------------------------------------------------------------

df = df.dropna()

# --------------------------------------------------------------
# 6. Basic statistics
# --------------------------------------------------------------

print("\nNumber of earthquakes:", len(df))

print("\nMagnitude statistics:")
print(df["mag"].describe())

print("\nDepth statistics (km):")
print(df["depth"].describe())

print("\nDate range:")
print(df["time"].min())
print(df["time"].max())

# --------------------------------------------------------------
# 7. Preview cleaned data
# --------------------------------------------------------------

print("\nCleaned dataset:")
print(df.head())
```
